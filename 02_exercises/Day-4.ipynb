#### 1. Create a class called `Book` that has the attributes title, author, pages, price. It also has the method get_price() which returns price, and set_price() to ensure price is numeric.

Then create two instances of this class:

b1:
* title -> 'Practical Programming'
* author -> 'Gries, Campbell, Montojo'
* pages -> 383
* price -> 50

b2:
* title -> 'Building a Career in Data Science'
* author -> 'Robinson, Nolis'
* pages -> 322
* price -> 40

```python
class Book:
    def __init__(self, title, author, pages, price):
        self.title = title
        self.author = author
        self.pages = pages
        self._price = price

    def describe(self):
        print(f"{self.title} by {self.author}. {self.pages} pages, Costs ${self._price}")

    def get_price(self):
        return self._price
    
    def set_price(self, new_price):
        if type(new_price) in [int, float]:
            self._price = new_price
        else:
            print("Invalid price! Price must be numeric.")
```

```python
b1 = Book("Practical Programming", "Gries, Campbell, Montojo", 383, 50)
b1.describe()

b2 = Book("Building a Career in Data Science", "Robinson, Nolis", 322, 40)
b2.describe()
```

```python
b1.set_price(50)
b1.set_price("50 dollars")
b1.get_price()
```

#### 2. Create a `Phone` class that has the attributes model, id, name, brand and price. Create the getters and setters for each attribute. As well as the following methods: calculate_total which takes the price and calculates it by the tax argument.

```python
class Phone:
    def __init__(self, model, id, name, brand, price):
        self._model = model
        self._id = id
        self._name = name
        self._brand = brand
        self._price = price

    ### Getters
    def get_model(self):
        return self._model

    def get_id(self):
        return self._id

    def get_name(self):
        return self._name

    def get_brand(self):
        return self._brand

    def get_price(self):
        return self._price
    
    ### Setters
    def set_model(self, new_model):
        self._model = new_model

    def set_id(self, new_id):
        self._id = new_id

    def set_name(self, new_name):
        self._name = new_name

    def set_brand(self, new_brand):
        self._brand = new_brand

    def set_price(self, new_price):
        # Checking if the value is numeric before updating
        if type(new_price) in [int, float]:
            self._price = new_price
        else:
            print("Invalid price! Price must be numeric.")

    def calculate_total(self, price, tax):
        # Calculates total by multiplying price by (1 + tax rate)
        total = price * (1 + tax)
        return round(total, 2)
```

```python
phone1 = Phone("17 Pro Max", "ABC123", "iPhone", "Apple", 1500)
phone1.calculate_total(phone1.get_price(), 0.13)
```
