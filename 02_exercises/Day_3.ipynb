#### 1. Write a conditional that prints different messages if a bank account balance is below 3,000, between 3,000 and 10,000, or over 10,000.

```python
def account_balance(bank_account):
  """
  Prints different messages if a bank account balance is below 3,000, between 3,000 and 10,000, or over 10,000.
  """
  bank_account = int(bank_account)

  if bank_account < 3000:
    print("Your bank account balance is below $3,000.")
  elif bank_account <= 10000:
    print("Your bank account balance is between $3,000 and $10,000.")
  else:
    print("Your bank account balance is over $10,000.")
```

```python
account_balance("4030")
account_balance(2980)
```

#### 4. Write a function called `dict_intersect` that takes two dictionaries, d1 and d2, as arguments and returns a set that contains only the keys found in both of the original dictionaries.

```python
def dict_intersect(d1, d2):
  """
  Takes two dictionaries, d1 and d2, as parameters and returns a set that contains only the keys found in both of the original dictionaries.
  """

  set_d1 = set(d1.keys())
  set_d2 = set(d2.keys())

  return set_d1.intersection(set_d2)
```

```python
d1 = {
    "Canada" : "Ottawa",
    "United States" : "Washington D.C.",
    "Mexico" : "Mexico City"
}

d2 = {
    "Canada" : "Ottawa",
    "France" : "Paris",
    "Spain" : "Madrid"
}

dict_intersect(d1, d2)
```
