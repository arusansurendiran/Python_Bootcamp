## Strings

#### 1. Given string contains a combination of the lower and upper case letters. Write a program to arrange the characters of a string so that all lowercase letters should come first.
- Ex. PyNaTive --> yaivePNT

*Hint! Iterate each character from a string and check if the current character is the lower or upper case using `islower()` string function*

```python
def arrange_lower(string):
  """
  Arranges the characters of a string so that all lowercase letters come first.

  Parameters: string: A combination of lower and upper case letters
  """
  n = len(string)
  lower = ""
  upper = ""

  for i in range(n):
    if string[i].islower():
      lower += string[i]
    else:
      upper += string[i]

  return lower + upper
```

```python
arrange_lower("PyNaTive")
```

```python
arrange_lower("ASarusan")
```

#### 2. Write a program to check if two strings are balanced. For example, strings s1 and s2 are balanced if all the characters in the s1 are present in s2. The character’s position doesn’t matter.

- Ex. s1 = "Yn" , s2 = "PYnative" True, since Yn is in s2
- Ex. s1 = "Ynf", s2 = "PYnative" False, since f is not in s2

*Hint! Iterate each character from a string s1 and check if the current character is present in the string s2.*

```python
def balanced_string(s1, s2):
  """
  Check if two strings are balanced.

  Paramters: s1, s2: Both phrases in strings

  >>> balanced_string("Yn", "PYnative")
  True
  >>> balanced_string("Ynf", "PYnative")
  False
  """

  n = len(s1)

  new_s1 = ""

  for i in range(n):
    if s1[i] not in s2:
      return False
  return True
```

```python
## Another idea to write the function

def balanced_string_2(s1, s2):
  """
  Check if two strings are balanced.

  Paramters: s1, s2: Both phrases in strings
  """

  n = len(s1)

  new_s1 = ""

  for i in range(n):
    if s1[i] in s2:
      new_s1 += s1[i]

  return s1 == new_s1
```

```python
string1 = "Yn"
string2 = "PYnative"

print(balanced_string(string1, string2))
print(balanced_string_2(string1, string2))
```

```python
string1 = "Ynf"
string2 = "PYnative"

print(balanced_string(string1, string2))
print(balanced_string_2(string1, string2))
```

#### 1. Write a Python program that asks the user for their name using the `input()` function and then greets them with a personalized message. Ensure that the program displays the user's name exactly as entered.

```python
def hello_there():
  """
  Asks the user for their name and then greets them with a personalized message.
  """
  name = input("What is your name?")
  print(f"Hey {name}! It's great to meet you. Welcome to the world of Python. Feel free to ask questions.")
```

#### 2. Write a program to accept two integer numbers from the user and calculate their product (multiplication).

```python
def multiply():
  """
  Accept two integer numbers from the user and calculate their product.
  """
  first_int = int(input("What's your first integer"))
  second_int = int(input("What's your second integer"))

  int_product = first_int * second_int

  print(f"Well, given your two integers, {first_int} x {second_int} = {int_product}")
```

```python
multiply()
```
